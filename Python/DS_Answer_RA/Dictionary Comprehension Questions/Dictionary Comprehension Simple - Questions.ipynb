# Dictionary Comprehension Simple Questions

```python
# 1. Write a Python program to create a simple dictionary using dictionary comprehension.
# Data:
#     range(5)
# Expected Output:
#     {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

result = {x: x * x for x in range(5)}
print(result)
```

```python
# 2. Write a Python program to iterate over dictionaries using for loops.
# Data:
#     d = {"a":10,"b":20,"c":30,}
# Expected Output:
#     {('a', 10): {'a': 10, 'b': 20, 'c': 30}, ('b', 20): {'a': 10, 'b': 20, 'c': 30}, ('c', 30): {'a': 10, 'b': 20, 'c': 30}}

d = {"a": 10, "b": 20, "c": 30}
result = {item: d for item in d.items()}
print(result)
```

```python
# 3. Write a Python program to display the following output using dictionary comprehension, that uses string as an iterable.
# Data:
#     string = "RED"
# Expected Output:
#     {'R': 'RRR', 'E': 'EEE', 'D': 'DDD'}

string = "RED"
result = {char: char * 3 for char in string}
print(result)
```

```python
# 4. Write a python program using dictionary comprehension to create a dictionary from the given list.
# The keys are the lower cases of each list items and the values are the upper cases of each list items.
# Data:
#     X = ['ReD', 'GrEeN', 'BlUe']
# Expected Output:
#     {'red': 'RED', 'green': 'GREEN', 'blue': 'BLUE'}

X = ['ReD', 'GrEeN', 'BlUe']
result = {item.lower(): item.upper() for item in X}
print(result)
```

```python
# 5. Write a Python program to extract particular keys from a dictionary using dictionary comprehension.
# Data:
#     D = {0: 'A', 1: 'B', 2: 'C', 3: 'D', 4: 'E', 5: 'F'}
#     Keys = [0, 2, 5]
# Expected Output:
#     {0: 'A', 2: 'C', 5: 'F'}

D = {0: 'A', 1: 'B', 2: 'C', 3: 'D', 4: 'E', 5: 'F'}
Keys = [0, 2, 5]
result = {key: D[key] for key in Keys}
print(result)
```

```python
# 6. Write a Python program to extract particular keys from a dictionary.
# Display all other items in a dictionary except the extracted keys using dictionary comprehension.
# Data:
#     D = {0: 'A', 1: 'B', 2: 'C', 3: 'D', 4: 'E', 5: 'F'}
#     Keys = [0, 1, 4]
# Expected Output:
#     {2: 'C', 3: 'D', 5: 'F'}

D = {0: 'A', 1: 'B', 2: 'C', 3: 'D', 4: 'E', 5: 'F'}
Keys = [0, 1, 4]
result = {key: value for key, value in D.items() if key not in Keys}
print(result)
```

```python
# 7. Write a python program to retrieve key and value from a dictionary and print the reverse using dictionary comprehension.
# For example: {'k': 'v'} -> {'v': 'k'}
# Data:
#     D = {0: 'Boss', 1: 'Siva', 2: 'Suresh'}
# Expected Output:
#     {'Boss': 0, 'Siva': 1, 'Suresh': 2}

D = {0: 'Boss', 1: 'Siva', 2: 'Suresh'}
result = {value: key for key, value in D.items()}
print(result)
```

```python
# 8. Write a Python program to wrap the list in enumerate() and pass it as an iterable to dictionary comprehension.
# Data:
#     X = ['red', 'green', 'blue']
# Expected Output:
#     {0: 'red', 1: 'green', 2: 'blue'}

X = ['red', 'green', 'blue']
result = {index: value for index, value in enumerate(X)}
print(result)
```

```python
# 9. Write a Python program to convert the given list into a dictionary with default value 2 for each key using dictionary comprehension.
# Data:
#     keys = ['red', 'green', 'blue']
# Expected Output:
#     {'red': 2, 'green': 2, 'blue': 2}

keys = ['red', 'green', 'blue']
result = {key: 2 for key in keys}
print(result)
```

```python
# 10. From the given lists keys and values, write a python program using dictionary comprehension to combine them.
# Data:
#     keys = ['name', 'age', 'job']
#     values = ['Suresh', 27, 'Dev']
# Expected Output:
#     {'name': 'Suresh', 'age': 27, 'job': 'Dev'}

keys = ['name', 'age', 'job']
values = ['Suresh', 27, 'Dev']
result = {key: value for key, value in zip(keys, values)}
print(result)
```

```python
# 11. Write a python program to print a dictionary only when the values are even numbers.
# The dictionary should have the key as x and values as x*x for range(6).
# Data:
#     range(6)
#     x % 2 == 0
# Expected Output:
#     {0: 0, 2: 4, 4: 16}

result = {x: x * x for x in range(6) if x % 2 == 0}
print(result)
```

```python
# 12. Write a Python program to sort a list alphabetically present in a dictionary using dictionary comprehension.
# Data:
#     d = {'n1': [2, 3, 1], 'n2': [5, 1, 2], 'n3': [3, 2, 4]}
# Expected Output:
#     {'n1': [1, 2, 3], 'n2': [1, 2, 5], 'n3': [2, 3, 4]}

d = {'n1': [2, 3, 1], 'n2': [5, 1, 2], 'n3': [3, 2, 4]}
result = {key: sorted(value) for key, value in d.items()}
print(result)
```
