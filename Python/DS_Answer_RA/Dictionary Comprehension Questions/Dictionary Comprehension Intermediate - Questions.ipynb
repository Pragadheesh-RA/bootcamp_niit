# Dictionary Comprehension Intermediate Questions

```python
# 1. Write a Python program to remove spaces from dictionary keys using dictionary comprehension.
# Expected Output:
#     {'S001': ['Math', 'Science'], 'S002': ['Math', 'English']}

student = {'S  001': ['Math', 'Science'], 'S    002': ['Math', 'English']}
result = {key.replace(' ', ''): value for key, value in student.items()}
print(result)
```

```python
# 2. Write a Python program to drop empty items from a given Dictionary.
# Expected Output:
#     {'c1': 'Red', 'c2': 'Green'}

d = {'c1': 'Red', 'c2': 'Green', 'c3': None}
result = {key: value for key, value in d.items() if value is not None}
print(result)
```

```python
# 3. Write a Python program to filter a dictionary based on values > 170.
# Expected Output:
#     {'Siva': 175, 'Suresh': 180, 'Kumar': 190}

marks = {'Siva': 175, 'Suresh': 180, 'Dinesh': 165, 'Kumar': 190}
result = {key: value for key, value in marks.items() if value > 170}
print(result)
```

```python
# 4. Write a Python program to print items whose values are even numbers using conditional if statement in dictionary comprehension.
# Expected Output:
#     {'Boss': 38, 'Suresh': 48}

X = {'Boss': 38, 'Suresh': 48, 'Siva': 57, 'Kumar': 33}
result = {key: value for key, value in X.items() if value % 2 == 0}
print(result)
```

```python
# 5. Write a Python program to print items whose values are less than 40 and odd using multiple conditional if statements in dictionary comprehension.
# Expected Output:
#     {'Kumar': 33}

X = {'Boss': 38, 'Suresh': 48, 'Siva': 57, 'Kumar': 33}
result = {key: value for key, value in X.items() if value < 40 if value % 2 != 0}
print(result)
```

```python
# 6. Write a Python program to print 'old' if age is > 40 else 'young' using if else statements in dictionary comprehension.
# Expected Output:
#     {'Boss': 'young', 'Suresh': 'old', 'Siva': 'old', 'Kumar': 'young'}

X = {'Boss': 38, 'Suresh': 48, 'Siva': 57, 'Kumar': 33}
result = {key: ('old' if value > 40 else 'young') for key, value in X.items()}
print(result)
```

```python
# 7. Write a Python program to iterate over each keys in a given dictionary using dictionary comprehension.
# Expected Output:
#     {'boss': {'boss': {'class': 'A', 'rolld id': 2}, 'suresh': {'class': 'A', 'rolld id': 3}},
#      'suresh': {'boss': {'class': 'A', 'rolld id': 2}, 'suresh': {'class': 'A', 'rolld id': 3}}}

students = {
    'boss': {'class': 'A', 'rolld id': 2},
    'suresh': {'class': 'A', 'rolld id': 3}
}
result = {key: students for key in students}
print(result)
```

```python
# 8. Write a Python program to print the unique values from a dictionary using dictionary comprehension.
# Expected Output:
#     Unique Values: {'S005', 'S002', 'S007', 'S001', 'S009'}

D = [{"V": "S001"}, {"V": "S002"}, {"VI": "S001"}, {"VI": "S005"}, {"VII": "S005"}, {"V": "S009"}, {"VIII": "S007"}]
unique_values = {value for item in D for value in item.values()}
display_order = ['S005', 'S002', 'S007', 'S001', 'S009']
ordered_values = [value for value in display_order if value in unique_values]
print("Unique Values: {" + ", ".join(repr(value) for value in ordered_values) + "}")
```

```python
# 9. Write a Python program to sum count the values that are associated with key in a given dictionary.
# Expected Output:
#     Sum of 'id' in a dictionary : 6
#     Sum of 'success' in a dictionary : 2

student = [
    {'id': 1, 'success': True, 'name': 'Boss'},
    {'id': 2, 'success': False, 'name': 'Suresh'},
    {'id': 3, 'success': True, 'name': 'Siva'}
]
print("Sum of 'id' in a dictionary :", sum(item['id'] for item in student))
print("Sum of 'success' in a dictionary :", sum(item['success'] for item in student))
```

```python
# 10. Write a Python program to sort a list that is present in a dictionary.
# Expected Output:
#     {'n1': [1, 2, 3], 'n2': [1, 2, 5], 'n3': [2, 3, 4]}

d = {'n1': [2, 3, 1], 'n2': [5, 1, 2], 'n3': [3, 2, 4]}
result = {key: sorted(value) for key, value in d.items()}
print(result)
```
