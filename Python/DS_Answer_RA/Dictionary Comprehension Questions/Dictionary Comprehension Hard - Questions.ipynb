# Dictionary Comprehension Hard Questions

```python
# 1. Write a Python program to convert more than one list into nested dictionary using dictionary comprehension.
# Expected Output:
#     The nested dictionary : [{'Teddy': {'ratings': 5}}, {'Maara': {'Ticket': 6}}, {'Master': {'Collection': 7}}]

l1 = ["Teddy", 'Maara', 'Master']
l2 = ['ratings', 'Ticket', 'Collection']
l3 = [5, 6, 7]
result = [{key1: {key2: value}} for key1, key2, value in zip(l1, l2, l3)]
print("The nested dictionary :", result)
```

```python
# 2. Write a Python program to filter the height > 6 and weight >= 70 of students, which are stored in a dictionary.
# Expected Output:
#     Height > 6ft and Weight> 70kg:
#     {'dinesh': (6.2, 70)}

students = {'dinesh': (6.2, 70), 'Gokul': (5.9, 65), 'Ajay': (6.0, 68), 'Vijay': (5.8, 66)}
result = {name: details for name, details in students.items() if details[0] > 6 and details[1] >= 70}
print("Height > 6ft and Weight> 70kg:")
print(result)
```

```python
# 3. Write a Python program to extract a list of values whose key is 'Math' from a given list of dictionaries.
# Expected Output:
#     Extract a list of values from said list of dictonaries where subject =  Math[90, 89, 92]

students = [{'Math': 90, 'Science': 92}, {'Math': 89, 'Science': 94}, {'Math': 92, 'Science': 88}]
subject = 'Math'
result = [student[subject] for student in students]
print(f"Extract a list of values from said list of dictonaries where subject =  {subject}{result}")
```

```python
# 4. Write a Python program to filter even numbers from list of values stored in a given dictionary values.
# Expected Output:
#     Filter even numbers from said dictionary values:
#     {'V': [4, 6, 10], 'VI': [4, 12], 'VII': [8]}
#     {'V': [], 'VI': [], 'VII': [2]}

students1 = {'V': [1, 4, 6, 10], 'VI': [1, 4, 12], 'VII': [1, 3, 8]}
students2 = {'V': [1, 3, 5], 'VI': [1, 5], 'VII': [2, 7, 9]}

def filter_even_numbers(data):
    return {key: [number for number in values if number % 2 == 0] for key, values in data.items()}

print("Filter even numbers from said dictionary values:")
print(filter_even_numbers(students1))
print(filter_even_numbers(students2))
```

```python
# 5. Write a Python program to print the keys which has the shortest list of values in a given dictionary.
# Expected Output:
#     Shortest list of values with the keys of the said dictionary:
#     ['VI', 'VIII', 'X']

d = {'V': [10, 12], 'VI': [10], 'VII': [10, 20, 30, 40], 'VIII': [20], 'IX': [10, 30, 50, 70], 'X': [80]}
min_length = min(len(values) for values in d.values())
result = [key for key, values in d.items() if len(values) == min_length]
print("Shortest list of values with the keys of the said dictionary:")
print(result)
```

```python
# 6. Write a Python program to extract values from a given list of dictionaries and create a list of lists from those extracted values.
# Expected Output:
#     List of lists from given list of dictionary:
#     [[1, 'Chandler', 'V'], [2, 'Monica', 'IV'], [3, 'Ross', 'VI'], [4, 'Rachel', 'VI'], [5, 'Phoebe', 'VII'], [6, 'Joey', 'VIII']]

students = [
    {'student_id': 1, 'name': 'Chandler', 'class': 'V'},
    {'student_id': 2, 'name': 'Monica', 'class': 'IV'},
    {'student_id': 3, 'name': 'Ross', 'class': 'VI'},
    {'student_id': 4, 'name': 'Rachel', 'class': 'VI'},
    {'student_id': 5, 'name': 'Phoebe', 'class': 'VII'},
    {'student_id': 6, 'name': 'Joey', 'class': 'VIII'}
]
result = [[student['student_id'], student['name'], student['class']] for student in students]
print("List of lists from given list of dictionary:")
print(result)
```

```python
# 7. Write a Python program to convert a given list of lists into a dictionary.
# Expected Output:
#     {1: ['Basakr', 'V'], 2: ['Suresh', 'V'], 3: ['Siva', 'VI'], 4: ['Karthi', 'VI'], 5: ['Dinesh', 'VII']}

students = [[1, 'Basakr', 'V'], [2, 'Suresh', 'V'], [3, 'Siva', 'VI'], [4, 'Karthi', 'VI'], [5, 'Dinesh', 'VII']]
result = {student[0]: student[1:] for student in students}
print(result)
```

```python
# 8. Write a python program to update the list of values in a given dictionary.
# Expected Output:
#     {'Math': [89, 90, 91], 'Physics': [90, 92, 87], 'Chemistry': [90, 87, 93]}

d = {'Math': [88, 89, 90], 'Physics': [92, 94, 89], 'Chemistry': [90, 87, 93]}
result = {
    key: [mark + 1 for mark in value] if key == 'Math' else [mark - 2 for mark in value] if key == 'Physics' else value
    for key, value in d.items()
}
print(result)
```

```python
# 9. Write a Python program to sort by value from a list of nested dictionary.
# Expected Output:
#     [{'key': {'subkey': 10}}, {'key': {'subkey': 5}}, {'key': {'subkey': 1}}]

D = [{'key': {'subkey': 1}}, {'key': {'subkey': 10}}, {'key': {'subkey': 5}}]
result = sorted(D, key=lambda item: item['key']['subkey'], reverse=True)
print(result)
```

```python
# 10. Write a Python Program to demonstrate converting nested dictionary to mapped tuple using list comprehension and generator expression.
# Expected Output:
#     The grouped dictionary : [('x', (5, 1, 8)), ('y', (6, 4, 3))]

d = {'gfg': {'x': 5, 'y': 6}, 'is': {'x': 1, 'y': 4}, 'best': {'x': 8, 'y': 3}}
result = [(key, tuple(inner[key] for inner in d.values())) for key in next(iter(d.values()))]
print("The grouped dictionary :", result)
```
