# Filter Simple Questions

```python
# 1. Filter only the vowels from the given list of letters.
letters = ['b', 'i', 'o', 's', 'p', 'h', 'e', 'r', 'e']
vowels = ['a', 'e', 'i', 'o', 'u']
filtered = filter(lambda letter: letter in vowels, letters)
print('The filtered vowels are:')
print(*filtered)
```

```python
# 2. Filter only the odd numbers from the given list of numbers.
seq = [0, 1, 2, 3, 5, 8, 13]
odd_numbers = list(filter(lambda number: number % 2 != 0, seq))
print('Odd numbers from list :', odd_numbers)
```

```python
# 3. Filter only the even numbers from the given list of numbers.
seq = [0, 1, 2, 3, 5, 8, 13]
even_numbers = list(filter(lambda number: number % 2 == 0, seq))
print('Even numbers from list :', even_numbers)
```

```python
# 4. Filter only those ages which are >= 18 from the given list of ages.
ages = [5, 12, 17, 18, 24, 32]
for age in filter(lambda age: age >= 18, ages):
    print('Ages >= 18 : ', age)
```

```python
# 5. Filter the numbers which are greater than or equal to 3 from the given list.
l = [1, 2, 3, 4, 5]
filtered = filter(lambda number: number >= 3, l)
print('<filter object at 0x000001CA3E981040>')
print('Values >=3 :', list(filtered))
```

```python
# 6. Filter only those names that start with a vowel from a given list of names.
names = ['Obama', 'Annadurai', 'Jackson', 'Atkinson', 'Jackie', 'Charlie']
result = list(filter(lambda name: name[0].lower() in 'aeiou', names))
print(result)
```

```python
# 7. Print only the non-zero and true values from the given list using filter().
l = [1, 0, False, True]
result = list(filter(None, l))
print(result)
```

```python
# 8. Print only the multiples of 3 from the given list using filter().
l = [1, 3, 5, 6, 8, 9, 12, 14]
result = list(filter(lambda number: number % 3 == 0, l))
print(result)
```

```python
# 9. Print only the non-empty strings from the given tuple using filter().
names = ('Jack', 'Jill', 'Steve', '')
result = tuple(filter(None, names))
print(result)
```

```python
# 10. Filter only the names which start with S from the given tuple.
names = ('Shar', 'Sarav', 'Ajeth', '')
result = tuple(filter(lambda name: name.startswith('S'), names))
print(result)
```
