# For Loop (Intermediate) - Questions

```python
# 1. Count occurrences of 4.
def count_four(values):
    count = 0
    for value in values:
        if value == 4:
            count += 1
    return count

print(count_four([1, 4, 6, 7, 4]))
print(count_four([1, 4, 6, 4, 7, 4]))
```

```python
# 2. Check whether specified value is present in list.
def contains(value, values):
    for item in values:
        if item == value:
            return True
    return False

print(contains(3, [1, 5, 8, 3]))
print(contains(-1, [1, 5, 8, 3]))
```

```python
# 3. Create a histogram.
def histogram(values):
    for value in values:
        print('*' * value)

histogram([2, 3, 12, 5])
```

```python
# 4. Concatenate all elements in a list into a string.
values = [1, 5, 12, 2]
result = ''
for value in values:
    result += str(value)
print(result)
```

```python
# 7. Hash a word.
word = 'culture'
code_map = {'c': 'C', 'u': '2', 'l': '1', 't': '0', 'r': '2', 'e': '0'}
coded_word = ''
for char in word:
    coded_word += code_map[char]
print('Enter the word to be hashed: culture')
print('The coded word is:', coded_word)
```

```python
# 10. Find maximum and minimum numbers from a list.
seq = [0, 10, 15, 40, -5, 42, 17, 28, 75]
maximum = seq[0]
minimum = seq[0]
for value in seq:
    if value > maximum:
        maximum = value
    if value < minimum:
        minimum = value
print((maximum, minimum))
```

```python
# 11. Find a distinct pair of numbers whose product is odd.
def odd_product_pair(values):
    for i in range(len(values)):
        for j in range(i + 1, len(values)):
            if values[i] != values[j] and (values[i] * values[j]) % 2 != 0:
                return True
    return None

print(odd_product_pair([2, 4, 6, 8]))
print(odd_product_pair([1, 6, 4, 7, 8]))
```

```python
# 14. Find number of notes.
def count_notes(amount):
    notes = [500, 200, 100, 50, 20, 10]
    count = 0
    for note in notes:
        count += amount // note
        amount %= note
    return count

print(count_notes(880))
print(count_notes(1000))
```

```python
# 15. Find the number of divisors of a given integer.
def divisor_count(number):
    count = 0
    for value in range(1, number + 1):
        if number % value == 0:
            count += 1
    return count

for number in [15, 12, 9, 6, 3]:
    print(divisor_count(number))
```

```python
# 16. Get the smallest number from a list.
def smallest(values):
    result = values[0]
    for value in values:
        if value < result:
            result = value
    return result

print(smallest([1, 2, -8, 0]))
```

```python
# 17. Get the largest number from a list.
def largest(values):
    result = values[0]
    for value in values:
        if value > result:
            result = value
    return result

print(largest([1, 2, -8, 0]))
```
