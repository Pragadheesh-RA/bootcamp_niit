# For Loop (Simple) - Questions

```python
# 1. Print every country in the list.
country = ['India', 'Canada', 'Sweden']
for item in country:
    print(item)
```

```python
# 2. Print every character in the string.
string = 'banana'
for char in string:
    print(char)
```

```python
# 3. Print output using break statement.
country = ['India', 'Canada', 'Sweden']
for item in country:
    if item == 'Sweden':
        break
    print(item)
```

```python
# 4. Print output using continue statement.
country = ['India', 'Canada', 'Sweden']
for item in country:
    if item == 'Canada':
        continue
    print(item)
```

```python
# 5. Empty for loop using pass statement.
for _ in []:
    pass
```

```python
# 6. Print expected output using range in for loop.
for number in range(6):
    print(number)
```

```python
# 7. Print expected output by passing range parameters as (2, 6).
for number in range(2, 6):
    print(number)
```

```python
# 8. Print range between 2 to 30 and increment by 3.
for number in range(2, 30, 3):
    print(number)
```

```python
# 9. Print all numbers from 0 to 5, and a message when loop ends.
for number in range(6):
    print(number)
print('Finally finished!')
```

```python
# 10. Print each adjective for every fruit using nested for loop.
adjective = ['red', 'big', 'tasty']
fruits = ['apple', 'banana', 'cherry']
for adj in adjective:
    for fruit in fruits:
        print(adj, fruit)
```

```python
# 11. Convert list of tuples into dictionary using setdefault().
l = [('x', 1), ('x', 2), ('x', 3), ('y', 1), ('y', 2), ('z', 1)]
result = {}
for key, value in l:
    result.setdefault(key, []).append(value)
print(result)
```

```python
# 12. Count elements until tuple appears.
x = [10, 20, 30, (10, 20), 40]
count = 0
for item in x:
    if isinstance(item, tuple):
        break
    count += 1
print(count)
```

```python
# 13. Access the index of elements in a list.
x = [5, 15, 35, 8, 98]
for index, value in enumerate(x):
    print(index, value)
```

```python
# 14. Create multiple lists.
result = {str(number): [] for number in range(1, 11)}
print(result)
```

```python
# 15. Create a list of empty dictionaries.
n = 5
result = [{} for _ in range(n)]
print(result)
```

```python
# 16. Iterate over two lists simultaneously.
x = [1, 2, 3]
name = ['boss', 'sai', 'bala']
for number, person in zip(x, name):
    print(number, person)
```

```python
# 17. Print all elements of the list using for loop and range().
name = ['boss', 'sai', 'bala']
for index in range(len(name)):
    print(name[index])
```

```python
# 18. Print without newline or space.
for char in 'Python':
    print(char, end='')
```

```python
# 19. Print expected output using enumerate().
x = ['suresh', 'boss', 'siva', 'bala', 'sai', 'balu', 'karthi']
for index, name in enumerate(x, start=1):
    print(f'President {index}: {name}')
```

```python
# 20. Print expected output using enumerate().
name = ['suresh', 'boss', 'siva', 'bala']
ratios = [0.2, 0.3, 0.5, 0.4]
for person, ratio in zip(name, ratios):
    print(f'{ratio * 100:.1f}% {person}')
```
